Make a 2 × 2 grid with one bar plot, one histogram, one box plot and one pie chart from this workbook.Give the whole figure a title and save it as a PNG.

Hint: plt.subplots(2, 2); fig.suptitle(...); plt.savefig('deals.png')

```python
import matplotlib.pyplot as plt
import seaborn as sns


# Create a 2 * 2 grid
fig, axes = plt.subplots(2, 2, figsize=(16, 12))


# Bar plot: Products per department
sns.countplot(data=df, x= "department", hue="department", palette="Set2", legend=False)

axes[0,0].set_title("Number of products per department")
axes[0,0].set_xlabel("Department")
axes[0,0].set_ylabel("Number of products")
axes[0,0].tick_params(axis="x", rotation = 45)

# Histogram: Discount percentage
sns.histplot(data=df, x="actual_discount_pct", bins=20, kde= True, ax= axes[0,1], color="steelblue")


axes[0, 1].set_title("Distribution of Discount Percentages")
axes[0, 1].set_xlabel("Discount (%)")
axes[0, 1].set_ylabel("Number of Products")


# 3. Box plot: Current prices by department
sns.boxplot(data=df, x="department", y="current_price_kes", ax=axes[1, 0], hue="department", palette="Pastel1", legend=False)

axes[1, 0].set_title("Current Prices by Department")
axes[1, 0].set_xlabel("Department")
axes[1, 0].set_ylabel("Current Price (KES)")
axes[1, 0].tick_params(axis="x", rotation=45)


# 4. Pie chart: Products by delivery type
delivery_counts = df["delivery_type"].value_counts()

axes[1, 1].pie(
    delivery_counts,
    labels=delivery_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

axes[1, 1].set_title("Products by Delivery Type")

# Give the entire figure a title
fig.suptitle(
    "Product Prices, Discounts and Delivery Analysis",
    fontsize=16,
    fontweight="bold"
)

# Adjust spacing
plt.tight_layout(rect=[0, 0, 1, 0.96])

# Save the figure as a PNG file
plt.savefig("deals.png", dpi=300, bbox_inches="tight")

# Display the figure
plt.show()
```

Final task · Mini deals report
Make 4 charts that answer the question “Where are the best Black November deals at Carrefour Kenya?”:
• a bar plot or horizontal bar plot comparing departments
• a histogram of discounts
• a box plot of discounts by department
• a pie chart or heatmap of your choice
Below the charts, write 3 sentences about what you found.
Checklist: every chart has a title · axes are labelled with units (KES, %) · labels don't overlap · colours are easy to

```python

import matplotlib.pyplot as plt
import seaborn as sns

# Keep only products with an actual discount
disc = df[df["actual_discount_pct"] > 0].copy()

# Calculate average discount by department
avg_discount = (
    disc.groupby("department")["actual_discount_pct"]
    .mean()
    .sort_values()
)

# Create a 2 × 2 figure
fig, axes = plt.subplots(2, 2, figsize=(16, 13))

# --------------------------------
# 1. Horizontal bar plot:
# Average discount by department
# --------------------------------
avg_plot = avg_discount.rename("avg_discount").reset_index()

sns.barplot(
    data=avg_plot,
    x="avg_discount",
    y="department",
    color="teal",
    ax=axes[0, 0]
)

axes[0, 0].set_title("Average Discount by Department")
axes[0, 0].set_xlabel("Average Discount (%)")
axes[0, 0].set_ylabel("Department")


# --------------------------------
# 2. Histogram:
# Distribution of discounts
# --------------------------------
sns.histplot(
    data=disc,
    x="actual_discount_pct",
    bins=20,
    color="cornflowerblue",
    edgecolor="white",
    ax=axes[0, 1]
)

axes[0, 1].set_title("Distribution of Discounts")
axes[0, 1].set_xlabel("Discount (%)")
axes[0, 1].set_ylabel("Number of Products")


# --------------------------------
# 3. Box plot:
# Discounts by department
# --------------------------------
sns.boxplot(
    data=disc,
    x="department",
    y="actual_discount_pct",
    color="lightblue",
    ax=axes[1, 0]
)

axes[1, 0].set_title("Discount Distribution by Department")
axes[1, 0].set_xlabel("Department")
axes[1, 0].set_ylabel("Discount (%)")
axes[1, 0].tick_params(axis="x", rotation=35)


# --------------------------------
# 4. Pie chart:
# Share of discounted products
# by department
# --------------------------------
department_counts = disc["department"].value_counts()

# Show the six largest departments and combine the rest
pie_counts = department_counts.head(6).copy()

if len(department_counts) > 6:
    pie_counts.loc["Other"] = department_counts.iloc[6:].sum()

wedges, texts, autotexts = axes[1, 1].pie(
    pie_counts.values,
    autopct="%1.1f%%",
    startangle=90,
    pctdistance=0.75
)

axes[1, 1].set_title("Share of Discounted Products by Department")

axes[1, 1].legend(
    wedges,
    pie_counts.index,
    title="Department",
    loc="center left",
    bbox_to_anchor=(1, 0.5)
)


# --------------------------------
# Three findings calculated from data
# --------------------------------
top_avg_dept = avg_discount.idxmax()
top_avg_value = avg_discount.max()

median_discount = disc["actual_discount_pct"].median()

top_count_dept = department_counts.idxmax()
top_count_share = (
    department_counts.max() / department_counts.sum()
) * 100

report = (
    f"1. {top_avg_dept} has the highest average discount at "
    f"{top_avg_value:.1f}%. "
    f"2. The median discount across discounted products is "
    f"{median_discount:.1f}%. "
    f"3. {top_count_dept} accounts for the largest share of "
    f"discounted products at {top_count_share:.1f}%."
)

# --------------------------------
# Overall title and layout
# --------------------------------
fig.suptitle(
    "Where Are the Best Black November Deals at Carrefour Kenya?",
    fontsize=17,
    fontweight="bold"
)

fig.subplots_adjust(
    top=0.90,
    bottom=0.20,
    hspace=0.45,
    wspace=0.35
)

# Add the three findings below the charts
fig.text(
    0.08,
    0.04,
    report,
    fontsize=10,
    wrap=True,
    verticalalignment="bottom"
)

# Save the complete report
plt.savefig(
    "carrefour_black_november_report.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# Print the findings as well
print(report)
```
